```python
import pickle
import os
import nest_asyncio

with open("/kaggle/input/datasets/rifahfatima/mynodes-pkl/nodes.pkl", "rb") as f:
    nodes = pickle.load(f)


nest_asyncio.apply()
```

```python

from llama_index.llms.huggingface import HuggingFaceLLM
import os
from dotenv import load_dotenv
from transformers import AutoModelForCausalLM, BitsAndBytesConfig

load_dotenv()

quants = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="float16",
)

model2 = HuggingFaceLLM(model_name="Qwen/Qwen2.5-3B-Instruct", tokenizer_name="Qwen/Qwen2.5-3B-Instruct",
                        model_kwargs={'quantization_config':quants}, device_map='auto',
                        context_window=3000,max_new_tokens=16)
```

```python
from llama_index.core.extractors import KeywordExtractor, QuestionsAnsweredExtractor
from llama_index.core.ingestion import IngestionPipeline

qa_extract = QuestionsAnsweredExtractor(questions=2, llm=model2)
keywords_extract = KeywordExtractor(keywords=5, llm=model2)
pipeline = IngestionPipeline(transformations=[keywords_extract, qa_extract])

for i in range(0, len(nodes), 10):
    now = nodes[i:i+10]
  

    print(f"processing {i} to next")
    extracted = pipeline.run(
    documents=now,
    show_progress=True)
    prevs=[]
    if os.path.exists("/kaggle/working/newextracts.pkl")==False:
        prevs = []
    else:
        with open("/kaggle/working/newextracts.pkl", "rb") as f: 
            prevs = pickle.load(f) 
    with open("/kaggle/working/newextracts.pkl", "wb") as f: 
        pickle.dump(prevs + extracted, f)
  
```
