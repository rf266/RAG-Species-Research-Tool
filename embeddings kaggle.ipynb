```python
import os
import pickle 
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
import nest_asyncio

nest_asyncio.apply()
```

```python
embed_model = HuggingFaceEmbedding(model_name="BAAI/bge-m3") #dense
```

```python
with open("/kaggle/input/datasets/rifahfatima/extracts-pkl/extracts.pkl", "rb") as f:
    extracted_nodes = pickle.load(f)
```

```python
embeds = []
for node in extracted_nodes:
    extracts = embed_model.get_text_embedding(node.get_content())
    embeds.append(extracts)
```

```python
with open("embeddings.pkl", "wb") as f:
    pickle.dump(embeds, f)
```
