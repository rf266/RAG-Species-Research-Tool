```python
from llama_index.core.node_parser import SemanticSplitterNodeParser
import pickle
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
import nest_asyncio
nest_asyncio.apply()
from dotenv import load_dotenv
import os
load_dotenv()
import pickle
from llama_index.llms.huggingface import HuggingFaceLLM
from transformers import BitsAndBytesConfig

embed_model = HuggingFaceEmbedding(
    model_name="BAAI/bge-small-en-v1.5"
)

quants = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="float16",
)

model2 = HuggingFaceLLM(model_name="Qwen/Qwen2.5-3B-Instruct", tokenizer_name="Qwen/Qwen2.5-3B-Instruct",
                        model_kwargs={'quantization_config':quants}, device_map='auto',
                        context_window=3000)


with open("/kaggle/input/datasets/rifahfatima/mds-pkl/mds.pkl", "rb") as f:
    docs = pickle.load(f)

     
```

```python
from llama_index.core.llms import ChatMessage
def custom_tag_classification(chunk):
    tags = ['paper_body', 'bibliography']
    prompt = f"This RAG pipeline requires the document corpus to be chunked. \n Each chunk has two classes: \n 0: paper_body text \n 1: bibliography \n to classify them. \n Here is the chunk: \n {chunk.text} \n You need to ONLY return 0 if its part of the body of the corpus,  paper_body, or return 1 if its a bibliography according to the attached output pydantic schema. \n Valid output can only be ONE character: 0 or 1"

    response = model2.chat([ChatMessage.from_str(prompt)])
    response = response.message.content.strip()
    if response == "0": 
        chunk.metadata["classification"] = tags[0]
    elif response=="1": 
        chunk.metadata["classification"] = tags[1]
    else: 
        print("invalid output")

    return chunk
    
```

```python
splitter = SemanticSplitterNodeParser(buffer_size=1, embed_model=embed_model,include_metadata=True)

nodes = []

for doc in docs:
    try: 
        chunks = splitter.get_nodes_from_documents([doc])
        for chunk in chunks:
            try:
                chunk = custom_tag_classification(chunk);
                nodes.append(chunk)
                print("chunk appended")
            except Exception as e:
                print(e, "unable to return chunk")
    except Exception as e:
        print(e, "Unable to chunk")
    
```

```python

with open("/kaggle/working/nodes.pkl", "wb") as f:
    pickle.dump(nodes, f)
```
